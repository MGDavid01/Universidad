---
Slicing 
tupla[start : stop : step]

Permite extraer una sección de la tupla especificando el índice inicial (start), el índice final **no incluyente (stop)** y el tamaño del paso (step).

````python
mediciones = (15.2, 18.0, 22.5, 25.1, 28.3, 30.0, 27.4, 21.0)

# Imprimir un rango específico [inicio:fin] (el límite 'fin' no se incluye)
print(mediciones[0:])

# Imprimir desde el inicio hasta el 5to elemento
print(mediciones[0:5])

# Imprimir las últimas 3 mediciones
print(mediciones[len(mediciones)-3:])

# Imprmir las mediciones de inicio a fin con paso de 2 
print(mediciones[0::2])

# Imprimir la tupla insertada
mediciones = mediciones + (11,)
print(mediciones[len(mediciones)-1])
````

---
```
zip(*iterables)
```
Combina varios iterables (listas, tuplas, etc.) elemento a elemento y devuelve un iterador de tuplas. Cada tupla contiene los elementos en la misma posición de cada iterable.


````python
# Combinar dos o más estructuras con zip
nombres = ('Ana', 'Luis', 'Maria')
edades = (22, 25, 20)
ciudades = ('Tijuana', 'Mexicali', 'Ensenada')

# zip genera tuplas (nombre, edad, ciudad)
combinados = tuple(zip(nombres, edades, ciudades))
print(combinados)

# Recorrer el resultado de zip
for nombre, edad, ciudad in zip(nombres, edades, ciudades):
    print(f'{nombre} tiene {edad} años y vive en {ciudad}')
````

Tareas a realizar:
1. Desempaqueta la tupla principal para extraer cada elemento en una variable
2. Imprime por separado el id, nombre, estado y cuantos registros tiene la estación.
2. Imprime cuántas veces se registró la temperatura de 38.0°C.
3. Imprime la primera posición donde ocurrió la temperatura máxima de la tupla.
4. Revisa si la temperatura crítica de 41.2°C está presente en los datos, de ser así, imprime un mensaje indicando su posición o posiciones.
5. Calcula e imprime la temperatura promedio, máxima y mínima
6. Haz una gráfica de las temperaruras 


````python
# Datos de entrada
datos_estacion = (104, "Tijuana_Este", (27.5, 28.0, 35.0, 38.0, 41.2, 38.0, 37.4, 38.0, 41.2), "Activa")

# Escribe tu código aquí
# 1. Desempaquetar la tupla principal
id_estacion, ubicacion, lecturas_temp, estado = datos_estacion

# 2. Imprimir por separado id, nombre, estado y cantidad de registros
print('ID:', id_estacion)
print('Nombre/Ubicacion:', ubicacion)
print('Estado:', estado)
print('Cantidad de registros:', len(lecturas_temp))

# 3. Cuántas veces se registró 38.0°C
print(f'Cantidad de veces para 38.0°C: {lecturas_temp.count(38.0)}')

# 4. Primera posición de la temperatura máxima
temp_max = max(lecturas_temp)
print(f'Posicion de la Primera Temperatura Maxima ({temp_max}°C): {lecturas_temp.index(temp_max)}')

# 5. Revisar si 41.2°C está presente e imprimir sus posiciones
temp_critica = 41.2
if temp_critica in lecturas_temp:
    posiciones = [i for i, t in enumerate(lecturas_temp) if t == temp_critica]
    print(f'Temperatura crítica {temp_critica}°C encontrada en las posiciones: {posiciones}')
else:
    print(f'Temperatura crítica {temp_critica}°C no encontrada.')

# 6. Temperatura promedio, máxima y mínima
promedio = sum(lecturas_temp) / len(lecturas_temp)
temp_min = min(lecturas_temp)
print(f'Temperatura Promedio: {promedio:.2f} | Temperatura Máxima: {temp_max} | Temperatura Mínima: {temp_min}')

# 7. Gráfica de temperaturas
import matplotlib.pyplot as plt

plt.plot(lecturas_temp, marker='o')
plt.title(f'Temperaturas - {ubicacion}')
plt.xlabel('Registro')
plt.ylabel('Temperatura (°C)')
plt.grid(True)
plt.show()
````

##### 4. Diccionarios
---
Estructura de datos mutable usada para almacenar elementos en pares, cada elemento tiene una llave (key) y un valor (value).

Los valores se consultan mediante su llave única (no por índice numérico).
Las llaves son inmutables.

````python
# Definición de un diccionario, utilizando {}
persona = {
  'Nombre': 'Sarai',
  'Edad': 27,
  'Documento': 1003882
}
print(persona)

# Acceso a un valor mediante su llave/Clave
print(persona['Nombre'])
print(persona['Documento'])

# Devuelve las claves
print(persona.keys())
# Devuelve los valores
print(persona.values())
# Devuelve las tuplas con los pares clave, valor
print(persona.items())
````

---
Constructor diccionario
```
dict()
```
Forma equivalente de crear un diccionario en Python es usando dict() e introduciendo los pares key: value entre paréntesis.

````python
persona = dict([
      ('Nombre', 'Sarai'),
      ('Edad', 27),
      ('Documento', 1003882),
])
print(persona)

persona2 = dict(Nombre = 'Laura', 
                Edad = 21,
                Documento = 1034320)
print(persona2)
````

---
Acceder y modificar un diccionario

Se puede acceder a sus elementos con [ ] o también con la función get( )
Si la clave no existe, .get() no rompe el programa con un error KeyError, sino que devuelve None (o un valor por defecto opcional que tú definas).

````python
# Creación del diccionario original
usuario = {'username': 'coder99', 'email': 'coder@mail.com', 'premium': False}

# Acceder mediante la llave
print(usuario['username'])
print(usuario.get('email'))
print(usuario.get('pass', None))

# Modificar un valor existente
usuario['premium'] = True
print(usuario)
````

---
Borrar elementos de un diccionario

```
del diccionario[llave]
```
La palabra **del**, borra un elemento específico mediante su llave. Si la llave no existe, regresa un error (keyError)


````python
# Creación del diccionario
motores = {'V6': 3.0, 'V8': 5.0, 'L4': 2.0}

# Elimina el motor V6 utilizando del
del motores['V6']
print(motores) 
````

```
pop()
```
El método **pop** elimina la llave especificada y devuelve su valor, este es útil cuando necesitas el valor del elemento eliminado. Para evitar errores si la llave no existe, puedes pasar un valor por default.

````python
# Creación del diccionario
usuario = {'username': 'coder99', 'email': 'coder@mail.com', 'premium': False}

# Eliminar mediante una clave
username_borrado = usuario.pop('usuario', None)
print(username_borrado)

email_borrado = usuario.pop('email')
print(usuario)
print(email_borrado)
````

```
popitem()
```
Elimina y devuelve el último par clave-valor que fue añadido al diccionario.

````python
# Creación del diccionario
carrito = {'Laptop': 5, 'Bocinas': 2, 'Teclado mecánico ': 3}

# Elimina el último elemento añadido
ultimo = carrito.popitem()

print(ultimo)   # Resultado: ('zapatos', 1)
print(carrito)  # Resultado: {'camisa': 1, 'pantalon': 2}
````

---
```
setdefault(llave, default)
```
Evita errores cuando consultamos una llave que no existe, con este método si la llave existe retorna su valor, si no existe, la crea y devuelve con un valor por default.

````python
configuracion = {'modo': 'oscuro'}

# Consulta llave que existe
modo_actual = configuracion.setdefault("modo", "claro")
print(configuracion)  


# Consulta una llave que no exite
volumen_actual = configuracion.setdefault('volumen', 60)
print(configuracion)
````

Este método ayuda a recorrer el diccionario con un ciclo, desempaqueta el diccionario en pares.

---
```
diccionario.copy()
```
Crea una copia superficial del diccionario.
```
diccionario.clear()
```
Elimina todos los elementos del diccionario.


````python
# Completar
# Diccionario original
datos_originales = {'a': 1, 'b': 2}

# crear una copia del diccionario
copia = datos_originales.copy()

# Limpiar el diccionario original
datos_originales.clear()

# Imprimir ambos diccionarios original y copia
print('Original:', datos_originales)
print('Copia:', copia)
````

````python
# Si el diccionario contiene elementos mutables anidados, y se modifican 
# ambos el diccionario original y la copia cambiarán su valor.

original = {'a': 1, 'b': [10, 20]}
print(original)
copia = original.copy()

# Modificar un elemento anidado afecta a ambos
copia['b'].append(30)
print(original['b'])  

print(original)
print(copia)
````

Ejercicio 3. Se tiene un diccionario con el estado actual de varios sensores. 
Algunos tienen errores de lectura (-99.0), otros están inactivos y algunos registros están incompletos.

Tareas a realizar:
1. Asegúrate de que todos los sensores tengan la clave "estado". Si no la tienen, asígnales "desconocido"
2. Saca una copia del diccionario, y elimina del diccionario los sensores inactivos, guarda sus datos en 'sensores_removidos'.
3. Actualiza el "sensor_3" para corregir su estado a "ok" y su batería a 50.
4. Recorre el diccionario e imprime solo los nombres de los sensores (llaves) que tienen una temperatura válida.
5. Extrae todas las temperaturas válidas usando .values() y calcula la temperatura promedio.
6. Haz una gráfica de las temperaturas válidas

````python
# Registro del estado de los sensores
datos_sensores = {
    'sensor_1': {'temp': 27.4, 'bateria': 85, 'estado': 'ok'},
    'sensor_2': {'temp': -99.0, 'bateria': 12, 'estado': 'error'},
    'sensor_3': {'temp': 29.8, 'bateria': 45},
    'sensor_4': {'temp': -99.0, 'bateria': 0, 'estado': 'inactivo'},
    'sensor_5': {'temp': 27.5, 'bateria': 75, 'estado': 'ok'},
    'sensor_6': {'temp': 30.0, 'bateria': 35, 'estado': 'ok'},
    'sensor_7': {'temp': -99.0, 'bateria': 0, 'estado': 'inactivo'},
    'sensor_8': {'temp': 26.3, 'bateria': 85, 'estado': 'ok'},
    'sensor_9': {'temp': 28.8, 'bateria': 65, 'estado': 'ok'},
    'sensor_10': {'temp': -99.0, 'bateria': 0, 'estado': 'inactivo'}
}

# Escribe tu código aquí

# 1. Asegurar que todos los sensores tengan la clave "estado"
for sensor in datos_sensores.values():
    sensor.setdefault('estado', 'desconocido')

print('1. Sensores con clave estado asegurada:')
for nombre, datos in datos_sensores.items():
    print(f'  {nombre}: {datos}')

# 2. Copia del diccionario; eliminar inactivos del diccionario de trabajo
datos_sensores = datos_sensores.copy()
sensores_removidos = {}

for nombre in list(datos_sensores.keys()):
    if datos_sensores[nombre]['estado'] == 'inactivo':
        sensores_removidos[nombre] = datos_sensores.pop(nombre)

print('\n2. Sensores removidos (inactivos):')
print(sensores_removidos)
print('Sensores restantes:')
print(list(datos_sensores.keys()))

# 3. Actualizar sensor_3: estado ok y batería 50
datos_sensores['sensor_3'].update({'estado': 'ok', 'bateria': 50})
print('\n3. sensor_3 actualizado:')
print(datos_sensores['sensor_3'])

# 4. Imprimir sensores con temperatura válida (distinta de -99.0)
print('\n4. Sensores con temperatura válida:')
for nombre, datos in datos_sensores.items():
    if datos['temp'] != -99.0:
        print(nombre)

# 5. Extraer temperaturas válidas usando .values() y calcular promedio
temperaturas_validas = []
for datos in datos_sensores.values():
    if datos['temp'] != -99.0:
        temperaturas_validas.append(datos['temp'])

promedio_temp = sum(temperaturas_validas) / len(temperaturas_validas)
print(f'\n5. Temperaturas válidas: {temperaturas_validas}')
print(f'   Temperatura promedio: {promedio_temp:.2f}°C')

# 6. Gráfica de las temperaturas válidas
import matplotlib.pyplot as plt

nombres_validos = [
    nombre for nombre, datos in datos_sensores.items() if datos['temp'] != -99.0
]

plt.bar(nombres_validos, temperaturas_validas)
plt.title('Temperaturas válidas de sensores')
plt.xlabel('Sensor')
plt.ylabel('Temperatura (°C)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.grid(axis='y')
plt.show()
````
