##### 1. Manejo de Strigs
---
Un string es una secuencia inmutable de caracteres. Hereda automáticamente el slicing, los índices y operaciones nativas.

````python
cadena = 'Hola mundo cruel'
print(cadena[0])
print(cadena[0:4])
````

```
strip():
```
Elimina los espacios en blanco (o caracteres especificados) al inicio y al final de una cadena.


````python
# Ejemplo de uso de la función strip() en Python
texto = '  Esto es un texto con espacios  '
texto_limpio = texto.strip()
print(texto_limpio)
# Ejemplo para eliminar caracteres específicos al inicio y al final de una cadena
texto2 = '###Hola Mundo##'
texto_limpio2 = texto2.strip('#')
print(texto_limpio2)
````

---
```
split(): 
```
Divide una cadena en una lista de subcadenas según un delimitador (por defecto, espacios).

````python
cadena = 'Python,Java,C++,JavaScript'
texto_dividido = cadena.split(',')
print(texto_dividido)
print(texto_dividido[0])
````

---
```
join(): 
```

Une los elementos de una lista de cadenas en una sola cadena, usando un separador.

````python
# Ejemplo de uso de la función join() en Python
lista_fecha = ['2026', '09', '07']
texto_unido = '-'.join(lista_fecha)
print(texto_unido)
letters = ['P', 'y', 't', 'h', 'o', 'n']
texto_unido2 = ''.join(letters)
print(texto_unido2)

````

---
f-strings

Permite incrustar variables o expresiones directamente dentro de una cadena anteponiendo la letra f.

````python
nombre = 'David'
experiencia = 11

print(f'Mi nombre es {nombre} y tengo {experiencia} años de experiencia')

# Ejemplo de uso de f-strings para formatear números
pi = 3.14159265

print(f'El valor de pi es {pi:.2f}')

producto = "Laptop"
precio = 1500

# Alinear a la izquierda (10 caracteres) y a la derecha (8 caracteres)
print(f"{producto:<10} | ${precio:>8.2f}")
# Output: Laptop     | $ 1500.00
````

**Ejercicio 1. Strings**

Desarrolla un script que procese la lista de datos de entrada y genere una nueva lista con los correos electrónicos normalizados (sin espacios, en minúsculas) y otra lista con los nombres de usuario extracted (todo antes del @).

Tareas a realizar:
1. Crear 'correos_limpios': lista con correos sin espacios al inicio/final y en minúsculas.
2. Crear 'nombre_usuarios': lista extrayendo solo la parte previa al símbolo '@'.


````python
# Datos de entrada
usuarios = ["  JUAN_PEREZ@gmail.com \n",
            "maria.Gomez@Hotmail.com  ",
            "  CARLOS_LOPA@yahoo.es\t",
            "ANA_torres@gmail.COM "
            ]

# --- Escribe tu código aquí ---
# Ejemplo de una linea de codigo para obtener los correos y usuarios
# lista_correos = [i.strip().lower() for i in usuarios]
# lista_usuarios = [i.split('@')[0] for i in usuarios]

# Ejemplo con mejor legibilidad
lista_correos = []
lista_usuarios = []

for i in usuarios:
    correo_limpio = i.strip().lower()
    correo_dividido = correo_limpio.split('@')
    lista_correos.append(correo_limpio)
    lista_usuarios.append(correo_dividido[0])

for j in range(len(lista_correos)):
    print(f'Correo: {lista_correos[j]:<25}| Usuario: {lista_usuarios[j]}')
````


##### 2. Estructuras de Datos y Operaciones
---
Listas

Estructuras de datos ordenadas y mutables que permiten almacenar varios elementos de cualquier tipo

---
Indexacion

Permite acceder a un elemento específico de la lista utilizando su posición (índice base 0). Los índices negativos cuentan desde el final.

````python
frutas = ['manzana', 'banana', 'cereza', 'durazno']


# Acceso negativo (de derecha a izquierda)
frutas_reversa = frutas[::-1]
print(frutas_reversa)
````

---
Slicing ( lista[start : stop : step] )

Permite extraer una sublista especificando el índice inicial (start), el índice final **no incluyente (stop)** y el tamaño del paso (step).

````python
numeros = [0, 10, 20, 30, 40, 50, 60, 70]

# Sublista del índice 1 al 4
numeros_sublista = numeros[1:4]
print(numeros_sublista)

# Sublista desde el inicio hasta el índice 5, paso de 2 en 2
numeros_sublista_2 = numeros[::2]
print(numeros_sublista_2)

# Invertir una lista completa usando paso negativo (-1)
numeros_invertidos = numeros[::-1]
print(numeros_invertidos)
````

Inserta un elemento en la posición indicada, los demas elementos se recorren.

````python
puntajes = [85, 90, 78, 92, 88]
puntajes.insert(0, 80)
print(puntajes)

# Ordenar la lista de puntajes de menor a mayor
for i in range(len(puntajes)):
    for j in range(len(puntajes)):
        if puntajes[i] < puntajes[j]:
            puntajes[i], puntajes[j] = puntajes[j], puntajes[i]
print(puntajes)
````

##### Eliminar elementos
---
```
pop(index)
```

Elimina y devuelve el elemento en el índice indicado (si no se le pasa parámetro, elimina y devuelve el último elemento).

````python
tasks = ['diseñar', 'codificar', 'probar', 'documentar', 'depurar']
print(tasks)
# Eliminar el último elemento
tasks.pop()
print(tasks)
tasks = ['diseñar', 'codificar', 'probar', 'documentar', 'depurar']
# Eliminar un elemento en un índice específico
tasks.pop(0)
print(tasks)
````

**Ejercicio 2. Filtrado y transformación de métricas**

Se tiene una lista de lecturas de temperatura en grados Celsius tomadas por un sensor durante el día (turno matutino y vespertino), cada hora se toma una muestra.


Tareas a realizar:
1. Imprime las lecturas correspondientes al horario vespertino utilizando slicing.
2. Divide en dos listas las lecturas del turno vespertino, y turno matutino
3. Elimina los valores de error (-99.0) y conserva solo las temperaturas válidas.
4. Obten la temperatura menor y mayor ordenando la lista de lecturas


````python

lecturas = [18.5, -99.0, 22.1, 25.4, -99.0, 30.2, 28.7, 15.0, -99.0, 20.5, 23.3, 25.7, 
            19.8, 21.4, -99.0, 24.6, 26.1, 27.5, -99.0, 29.3, 31.0, 32.5, 33.8, 34.2]


# --- Escribe tu código aquí ---

# Eliminar los valores de error
while -99.0 in lecturas:
    lecturas.remove(-99.0)
print(f'Lecturas sin errores: {lecturas} \n')
# Ordenar de menor a mayor
lecturas.sort()
print(f'Lecturas ordenadas: {lecturas} \n')
# Dividir en dos listas
turno_matutino = lecturas[:10]
turno_vespertino = lecturas[10:]
print(f'Turno matutino: {turno_matutino} | Turno vespertino: {turno_vespertino} \n')
# Obtener la temperatura menor y mayor
print(f'Temperatura menor: {lecturas[0]} | Temperatura mayor: {lecturas[-1]} \n')

import matplotlib.pyplot as plt

plt.figure(figsize=(10, 5))
plt.plot(turno_matutino, label='Turno Matutino')
plt.plot(turno_vespertino, label='Turno Vespertino')
plt.xlabel('Hora')
plt.ylabel('Temperatura (Celcius)')
plt.title('Temperatura por turno')
plt.legend()
plt.show()
````
