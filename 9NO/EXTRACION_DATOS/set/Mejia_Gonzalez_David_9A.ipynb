##### 5. Set
---
Permite almacenar varios elementos y acceder a ellos de una forma similar a las listas, sus diferencias principales son:
- Los elementos de un set son únicos, no puede haber elementos duplicados
- Los set son dessordenados, no mantienen el orden de cuando son declarados
- Sus elementos deben ser inmutables

---
Definición de un Set

````python
# Crear un set
lista = [1, 3, 5, 8, 2, 4, 8]
print(lista)

# Definir un set mediante su constructor
s = set(lista)
print(s)

# Se puede definir un set utilizando las {} 
set1 = {3, 4, 2, 9}
print(set1)
````

---
Operaciones con sets

````python
# Modificación de un set 
s = set([5, 6, 7, 8])
s[0] = 3 
````

````python
# Los elementos de un set deben ser inmutables
lista = ['Mexico', 'Argentina']
s = set(['Colombia', 'España', lista]) 
````

---
Iterar un set

````python
# Iterar un set de la misma forma que las listas
s = set([4, 7, 8, 2])
for elemento in s:
    print(elemento)
````

---
```
len()
```
Obtener la longitud de un set

````python
s = set([1, 2, 2, 3, 4, 1])
print(len(s))
````

---
```
in 
```
Conocer si un elemento esta presente en un set. Si el valor existe en el ser debolverá un True, de lo contrario un False.

````python
instrumentos = set(['Batería', 'Guitarra', 'Bajo'])
print('guitarra' in instrumentos)
````

---
Operador union |
```
union(otro set)
```

Devuelve un nuevo set con todos los elementos de ambos conjuntos.

````python
# operador union
s1 = set([1, 2, 3])
s2 = set([3, 4, 5])

print(s1 | s2) 

# Metodo union
print(s1.union(s2))
````

---
```
intersection(otro set)
```
Devuelve un conjunto con los elementos comunes que están en ambos sets.

````python
s1 = set([1, 2, 3, 6, 9])
s2 = set([3, 4, 5, 2])

# Metodo interseccion
print(s1.intersection(s2))
````

---
```
difference(otro set)
```
Devuelve los elementos que están en el primer set pero no en el segundo.

````python
s1 = set([1, 2, 3, 6, 9])
s2 = set([3, 4, 5, 2])

# Metodo diferencia
print(s1.difference(s2))
````

---
```
s.add(elemento)
```
Permite añadir elementos al set

````python
puntaje = set([2.5, 3.8, 5, 4])
puntaje.add(3)
print(puntaje)
````

---
```
s.remove(elemento)
```
Elimina el elemento que se pasa como parámetro. Si el elemento no se encuentra manda un error.

````python
s = set([3, 4, 1, 2])
s.remove(2)
print(s)
````

---
```
s.discard(elemento)
```
Borra el elemento del set y si no lo encuantra no hace nada.

````python
s = set([1, 2, 5])
s.discard(3)
print(s)
````

---
```
s.pop()
```
Elimina un elemento aleatorio del set.

````python
s1 = set([5, 4, 3, 2, 1])
s1.pop()
print(s1)
````

---
```
s.clear()
```
Elimina todos los elementos de un set.

````python
s1 = set([5, 4, 3, 2, 1])
s1.clear()
print(s1)
````

Ejercicio 1. Lsita de usuarios que ingresaron a una aplicación.
1. Convertir la lista de usuarios a un set para obetener usuarios únicos.
2. Mostrar cuantos usuarios diferentes hay y cuantos repetidos.
3. Agregar un nuevo usuario al set
4. Comprobar si el usuario Carlos ingresó a la aplicación.


````python
usuarios = ['Anabel', 'Carlos', 'Maria', 'Anabel', 'Pedro', 'Carlos', 'Luis', 'Maria', 'Ana']

set1 = set(usuarios)

print(f"Usuarios únicos: {len(set1)}")
print(f"Usuarios repetidos: {len(usuarios) - len(set1)}")

set1.add('Juan Carlos Bodoque')
print(f"Nota Verde: {set1}")

print(f"¿Carlos ingresó?: {'Carlos' in set1}")
````
