##### Pandas
Pandas es una librería de Python cuyo nombre deriva del término panel data (utilizado para referirse a conjuntos de datos estructurados multidimensionales). Proporciona estructuras de datos y funciones de alto nivel que permiten manipular y analizar datos estructurados de manera cómoda y eficiente.
Estructuras de datos principales:
- Serie (Series): Estructura unidimensional que contiene un arreglo de datos y un conjunto de etiquetas asociado llamado índice (index).
- DataFrame: Estructura tabular bidimensional orientada a columnas, con etiquetas organizadas en ejes verticales (filas/eje 0) y horizontales (columnas/eje 1).

---
Series

````python
# Importar libreria Pandas
import pandas as pd

# Construir una serie de Pandas a partir de una lista
datos = [100, 200, 300, 400, 500]
series= pd.Series(datos)

print(f'Series: \n{series}')

datos1 = [100, 200, 300.1, 400, 500]
series1= pd.Series(datos1)
print(f'\nSeries1: \n{series1}')


print(f'\nEl primer elemento de la serie es: {series[0]}')
````

````python
# Construir una serie de Pandas con una lista de índices personalizados
series_datos = pd.Series(datos, index=['sensor1', 'sensor2', 'sensor3', 'sensor4', 'sensor5'])
print(series_datos)
````

````python
# Creacion de una serie de Pandas a partir de un diccionario
produccion = {
    'Dia 1': 1000,
    'Dia 2': 2100,
    'Dia 3': 1500,
    'Dia 4': 1300,
    'Dia 5': 9000,
}

series_prod = pd.Series(produccion)
print(series_prod)
````

--- 
Acceder a elemenos de una serie
```
NombreSerie[index]

NombreSerie.loc[etiqueta]

NombreSerie.iloc[posicion]
```

````python
# Imprimir Serie de Pandas
print(series_prod)

# Acceder a un elemento de la serie de datos por su índice
print(f'\nEl primer elemento de la serie de datos es: {series_prod.iloc[0]}')

# Acceder a un elemento de la serie de datos por su etiqueta
print(f"\nDia 1: {series_prod['Dia 1']}")
````

````python
# Acceder a un elemento de la serie por su índice
print(f"\nIndice: {series_datos['sensor1']}")

# Loc: Usa los nombres de las filas o columnas en lugar de su posición numérica entera.
print(f"Loc: {series_datos.loc['sensor1']}")

# iloc: Usa la indexación por posición numérica entera.
print(f'Iloc: {series_datos.iloc[0]}')
````

---
Remplazar indices, agregar elementos

````python
# Remplazar los índices de la serie por otros personalizados
series_datos = pd.Series(datos, index=['ap1', 'ap2', 'ap3', 'ap4', 'ap5'])

print(series_datos)

print(series_datos.loc['ap1'])
print(series_datos.iloc[2])
````

````python
# Agregar un nuevo elemento a la serie
series_datos['ap6'] = 600

# Imprimir la serie modificada
print('\nSerie Modificada: ')
print(series_datos)
````

````python
# Filtrar elementos de la serie que cumplan una condición
print('\nSerie datos: ')
print(series_datos)

# Remplazar los índices de la serie por otros personalizados
series_datos.index = ['sensor1', 'sensor2', 'sensor3', 'sensor4', 'sensor5', 'sensor6']
print(series_datos)

# Imprimir los elementos de la serie que sean mayores a 300
print('\nElementos mayores a 300:')
print(series_datos[series_datos > 300])
````

---
Recorrer una serie

````python
# Creamos una serie de ejemplo
valores = pd.Series([10, 20, 30, 35, 40, 30], index=['a', 'b', 'c', 'd', 'e', 'f'])

# Imprimir la serie
print(valores)

# Imprimir los índices y valores de la serie que sean iguales a 30
print('\nElementos iguales a 30:')
for indice, valor in valores.items():
    if valor == 30:
        print(f'Índice: {indice}, Valor: {valor}')
````

Data Frame


````python
# Data frame
data = {
    'Nombre': ['Juan', 'Pedro', 'Maria', 'Ana'],
    'Edad': [25, 30, 28, 22],
    'Ciudad': ['Madrid', 'Barcelona', 'Valencia', 'Sevilla']
}

dataframe = pd.DataFrame(data)
print(dataframe)
````

````python
# Data frame con índices personalizados
dataframe = pd.DataFrame(data, index=['Persona 1', 'Persona 2', 'Persona 3', 'Persona 4'])
print(dataframe)
````

````python
dataframe = pd.DataFrame(data, index=['Persona 1', 'Persona 2', 'Persona 3', 'Persona 4'])
# Acceder a una columna del data frame por su índice
print(dataframe.loc['Persona 1'])

# Acceder a una columna del data frame por su posición
print(dataframe.iloc[1])
````

````python
# Acceder a un elemento específico del data frame por su posición en los índices y columnas
print(dataframe.iloc[0, 1])

# Acceder a un elemento específico del data frame por su índice y columna
print(dataframe.loc['Persona 1', 'Nombre'])
````

````python
# Modificar un elemento del data frame

# modificar los indices de la serie por otros personalizados
df_empleados = pd.DataFrame(data, index=['Empleado 1', 'Empleado 2', 'Empleado 3', 'Empleado 4'])
print('\n',df_empleados)

# Insertar una nueva columna en el data frame
df_empleados['Departamento'] = ['Ventas', 'IT', 'RRHH', 'Marketing']
print('\nCon nueva columna:\n', df_empleados)

# Agregar un nuevo registro al data frame
df_empleados.loc['Empleado 5'] = ['Carlos', 35, 'Bilbao', 'Finanzas']
print('\nCon nuevo registro:\n', df_empleados)

````

````python
# Atributos del data frame

print(f'Índices: {df_empleados.index}')

print(f'\nColumnas: {df_empleados.columns}')

print(f'\nEjes: {df_empleados.axes}')

print(f'\nForma: {df_empleados.shape}')
````

Ejercicios de filtrado

````python
# Filtrar todos los pokemones que tengan una altura mayor a 2.0 metros
# Mostrar la lista de pokemones que cumplan con la condición y mencionar cuantos son
altos = datos[datos['Altura'] > 2.0]
print(altos[['Nombre', 'Altura']])
print(f'\nCantidad de pokemones con altura mayor a 2.0 m: {len(altos)}')
````

````python
# Filtrar todos los pokemones que tengan un peso mayor a 100.0 kilogramos
# Mostrar la lista de pokemones que cumplan con la condición y mencionar cuantos son
pesados = datos[datos['Peso'] > 100.0]
print(pesados[['Nombre', 'Peso']])
print(f'\nCantidad de pokemones con peso mayor a 100.0 kg: {len(pesados)}')
````

````python
# Filtrar todos los pokemones Legendarios
# Mostrar la lista de pokemones que cumplan con la condición y mencionar cuantos son
legendarios = datos[datos['Legendario'] == 1]
print(legendarios[['Nombre', 'Tipo1', 'Tipo2', 'Legendario']])
print(f'\nCantidad de pokemones legendarios: {len(legendarios)}')
````

````python
# Filtrar todos los pokemones que sean de tipo1 Agua y Tipo2 Agua (Los operadores bit a bit son: & (and), | (or), ~ (not))
# Mostrar la lista de pokemones que cumplan con la condición y mencionar cuantos son
agua_agua = datos[(datos['Tipo1'] == 'Agua') & (datos['Tipo2'] == 'Agua')]
print(agua_agua[['Nombre', 'Tipo1', 'Tipo2']])
print(f'\nCantidad de pokemones Tipo1 Agua y Tipo2 Agua: {len(agua_agua)}')
````

````python
# Filtrar todos los pokemones que sean de tipo1 Fuego y Tipo2 Volador
# Mostrar la lista de pokemones que cumplan con la condición y mencionar cuantos son
fuego_volador = datos[(datos['Tipo1'] == 'Fuego') & (datos['Tipo2'] == 'Volador')]
print(fuego_volador[['Nombre', 'Tipo1', 'Tipo2']])
print(f'\nCantidad de pokemones Tipo1 Fuego y Tipo2 Volador: {len(fuego_volador)}')
````
